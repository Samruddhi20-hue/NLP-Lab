```python
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
```

```python
data = {
    "text": [
        "I love NLP! It is amazing.",
        "Machine learning is very useful.",
        "I hate bugs in code.",
        "Debugging code is frustrating but important."
    ],
    "label": ["positive", "positive", "negative", "negative"]
}

df = pd.DataFrame(data)

print("Original Data:\n", df)
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    text = text.lower()
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word.isalpha()]
    tokens = [word for word in tokens if word not in stop_words]
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return " ".join(tokens)
```

```python
df['cleaned_text'] = df['text'].apply(clean_text)
print("Data after text cleaning:\n", df)
```

```python
# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit and transform the cleaned text
X = tfidf_vectorizer.fit_transform(df['cleaned_text'])

print("TF-IDF features shape:", X.shape)
print("\nFirst 5 TF-IDF feature vectors (sparse matrix):")
print(X[:5])

# Initialize Label Encoder
label_encoder = LabelEncoder()

# Encode the labels
y = label_encoder.fit_transform(df['label'])

print("\nEncoded labels:", y)
print("Original labels:", df['label'].values)
print("Label mapping:", dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))
```

```python
le = LabelEncoder()
df["encoded_label"] = le.fit_transform(df["label"])

print("\nEncoded Labels:\n", df["encoded_label"])
```

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(df["cleaned_text"])

tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=tfidf.get_feature_names_out())

print("\nTF-IDF Matrix:\n", tfidf_df)
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
