```python
import nltk
nltk.download('punkt')

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from gensim.models import Word2Vec
import numpy as np
```

```python
corpus = [
    "I love natural language processing",
    "Natural language processing is very interesting",
    "I love machine learning and NLP"
]

print("Corpus:")
for sentence in corpus:
    print(sentence)
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print("\n--- Bag of Words (Count Occurrence) ---")
print("Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", X.toarray())
```

```python
bow_matrix = X.toarray()
normalized_bow = bow_matrix / bow_matrix.sum(axis=1, keepdims=True)

print("\n--- Normalized Bag of Words ---")
print(normalized_bow)
```

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)

print("\n--- TF-IDF Representation ---")
print("Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X_tfidf.toarray())
```

```python
nltk.download('punkt_tab')
tokenized_corpus = [nltk.word_tokenize(sentence.lower()) for sentence in corpus]
```

```python
model = Word2Vec(sentences=tokenized_corpus, vector_size=100, window=5, min_count=1)

print("\n--- Word2Vec Embeddings ---")
```

```python
print("\nVector for 'language':\n", model.wv['language'])
```

```python
print("\nWords similar to 'language':")
print(model.wv.most_similar('language'))
```

```python
print("\n--- SUMMARY ---")
print("BoW Shape:", X.shape)
print("TF-IDF Shape:", X_tfidf.shape)
print("Word2Vec Vocabulary:", list(model.wv.index_to_key))
```
