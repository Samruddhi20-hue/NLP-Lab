```python
from nltk.corpus import wordnet
```

```python

word = "good"
print("Input Word:", word)
```

```python
synonyms = set()

for syn in wordnet.synsets(word):
    for lemma in syn.lemmas():
        synonyms.add(lemma.name())

print("\n--- Synonyms ---")
print(synonyms)
```

```python
antonyms = set()

for syn in wordnet.synsets(word):
    for lemma in syn.lemmas():
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())

print("\n--- Antonyms ---")
print(antonyms)
```

```python
print("\n--- Hypernyms ---")

for syn in wordnet.synsets(word):
    hypernyms = syn.hypernyms()
    if hypernyms:
        print(f"{syn.name()} --> {[h.name() for h in hypernyms]}")
```

```python
text = "The dog is a good animal"
words = text.split()

print("\n--- Analysis for Each Word ---")

for w in words:
    syns = wordnet.synsets(w)
    if syns:
        print(f"\nWord: {w}")

        # Synonyms
        syn_list = set([lemma.name() for syn in syns for lemma in syn.lemmas()])
        print("Synonyms:", list(syn_list)[:5])

        # Antonyms
        ant_list = set()
        for syn in syns:
            for lemma in syn.lemmas():
                if lemma.antonyms():
                    ant_list.add(lemma.antonyms()[0].name())
        print("Antonyms:", list(ant_list))
```

```python
hyper_list = []
for syn in syns:
    for h in syn.hypernyms():
        hyper_list.append(h.name())
print("Hypernyms:", list(set(hyper_list))[:5])
```
