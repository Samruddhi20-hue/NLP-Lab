```python
import spacy
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Elon Musk, the CEO of Tesla, visited India in 2024.
He met Prime Minister Narendra Modi in New Delhi.
The meeting was held at the Rashtrapati Bhavan.
"""

print("Input Text:\n", text)
```

```python
doc = nlp(text)

print("\n--- Extracted Entities ---")
predicted_entities = []

for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}")
    predicted_entities.append((ent.text, ent.label_))
```

```python

true_entities = [
    ("Elon Musk", "PERSON"),
    ("Tesla", "ORG"),
    ("India", "GPE"),
    ("2024", "DATE"),
    ("Narendra Modi", "PERSON"),
    ("New Delhi", "GPE"),
    ("Rashtrapati Bhavan", "ORG")
]
```

```python
# Extract only labels
y_true = [label for (_, label) in true_entities]
y_pred = []

# Match predicted labels with true entities
for true_ent in true_entities:
    found = False
    for pred_ent in predicted_entities:
        if true_ent[0] == pred_ent[0]:
            y_pred.append(pred_ent[1])
            found = True
            break
    if not found:
        y_pred.append("O")  # Not detected

print("\nTrue Labels:", y_true)
print("Predicted Labels:", y_pred)
```

```python
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average='macro', zero_division=0)
recall = recall_score(y_true, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)

print("\n--- Evaluation Metrics ---")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```
