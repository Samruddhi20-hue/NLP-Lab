```python
from nltk.tokenize import word_tokenize, TreebankWordTokenizer, TweetTokenizer, MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
```

```python
text = "NLTK is a powerful library! It helps in Natural Language Processing tasks like tokenization, stemming, and lemmatization."

print("Original Text:\n", text)
```

```python
print("\n--- TOKENIZATION ---")
```

```python
whitespace_tokens = text.split()
print("\n1. Whitespace Tokenization:")
print(whitespace_tokens)
```

```python
import nltk
nltk.download('punkt_tab')
punctuation_tokens = word_tokenize(text)
print("\n2. Punctuation-based Tokenization:")
print(punctuation_tokens)
```

```python
treebank = TreebankWordTokenizer()
treebank_tokens = treebank.tokenize(text)
print("\n3. Treebank Tokenization:")
print(treebank_tokens)
```

```python
tweet = TweetTokenizer()
tweet_tokens = tweet.tokenize(text)
print("\n4. Tweet Tokenization:")
print(tweet_tokens)
```

```python
mwe_tokenizer = MWETokenizer([('Natural', 'Language')])
mwe_tokens = mwe_tokenizer.tokenize(word_tokenize(text))
print("\n5. MWE Tokenization:")
print(mwe_tokens)
```

```python
print("\n--- STEMMING ---")

porter = PorterStemmer()
snowball = SnowballStemmer("english")

tokens = word_tokenize(text)

porter_stems = [porter.stem(word) for word in tokens]
snowball_stems = [snowball.stem(word) for word in tokens]

print("\nPorter Stemmer Output:")
print(porter_stems)

print("\nSnowball Stemmer Output:")
print(snowball_stems)
```

```python
print("\n--- LEMMATIZATION ---")

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in tokens]

print("\nLemmatized Output:")
print(lemmatized_words)
```

```python
print("\n--- SUMMARY ---")
print("Original Tokens:", tokens)
print("Porter Stemmed:", porter_stems)
print("Snowball Stemmed:", snowball_stems)
print("Lemmatized:", lemmatized_words)
```
